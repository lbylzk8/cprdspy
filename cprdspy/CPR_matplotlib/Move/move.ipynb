# Move — matplotlib 版海龟绘图

**关于动画**：Jupyter 默认的 `inline` 后端没有事件循环，一个单元格只在结束时显示最后一帧，所以看起来像没有动画。

* 想在 notebook 里看到逐帧动画：装 `ipympl` 后把 `%matplotlib widget` 那行取消注释。
* 或者把同样的代码存成 `.py` 用 `python x.py` 运行（TkAgg / QtAgg 后端会真的动起来）。
* 只想快速看结果：`speed(0)` 关掉动画，瞬间画完。

```python
%matplotlib inline
# %matplotlib widget      # 想要逐帧动画就换成这一行（需要 ipympl）

import os
import sys

sys.path.insert(0, os.path.abspath('../../..'))   # 仓库内直接运行用

import matplotlib.pyplot as plt
from cprdspy.CPR_matplotlib.Move.move import Move

m = Move()
m.speed(0)          # 0 = 不做动画，瞬间画完
for _ in range(36):
    m.forward(120)
    m.left(170)
plt.show()
```

```python
# 也可以用 CPR_matplotlib 包里的 move_ 前缀接口，
# 并且能把海龟画到已有的 figure/axes 上（与其它 matplotlib 图形共存）
import matplotlib.pyplot as plt
import cprdspy.CPR_matplotlib as cpm

fig, ax = plt.subplots()
ax.plot([-300, 300], [0, 0], color='orange', linewidth=1)   # 普通 matplotlib 图形

t = cpm.move_Move(fig=fig, ax=ax)
t.speed(0)
t.forward(100)
t.left(90)
t.circle(50)

cpm.move_bgcolor('lightyellow')
cpm.move_forward(60)
cpm.move_update()
plt.show()
```

```python
# 默认颜色与数值精度检查
plt.close('all')
t2 = cpm.move_Move()
print('默认 pencolor:', t2.pencolor())        # (0.0, 1.0, 0.0) 即 #0f0
t2.speed(0)
t2.setheading(0)
t2.left(33.333333333333)
print('heading 精度:', t2.heading())          # 33.333333333333
print('pos 表示:', t2.pos())
```
