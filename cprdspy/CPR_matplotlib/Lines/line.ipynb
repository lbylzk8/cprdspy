```python
from cprdspy.CPR_matplotlib.Lines.line import *
from cprdspy.CPR_matplotlib.Dots.dot import *

dots = n_dots(6, R=1)

# 1. 统一颜色（str）
draw_lines(connect(dots), color="cyan")

# 2. 颜色名列表 —— 逐段指定
draw_lines(
    multi_polygon(6, m=3),
    color=["#ff0000", "#00ff00", "#0000ff", "#ff8800", "#ff00ff", "#00ffff"],
)

# 3. 数值列表 + cmap —— 渐变
draw_lines(connect_all(dots), color=[0, 1, 2, 3, 4, 5], cmap="rainbow")

# 4. color_by 自动渐变（不变）
draw_lines(connect_all(dots), color_by="a", cmap="hsv")
```
