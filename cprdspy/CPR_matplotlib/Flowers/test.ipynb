```python
from flower import flower_petal_fill, oval_petal_fill
import matplotlib.pyplot as plt

fig, (ax1, ax2) = plt.subplots(1, 2)

# 圆花瓣填充：红色填充 + 深红描边
flower_petal_fill(
    R=1,
    r=0.6,
    n=5,
    fill_color="#ff4444",
    fill_alpha=0.4,
    edge_color="#cc0000",
    edge_width=2,
    ax=ax1,
)

# 椭圆花瓣填充：蓝色填充 + 半透明
oval_petal_fill(
    a=2,
    b=1,
    d=0.3,
    rotation=45,
    fill_color="#4488ff",
    fill_alpha=0.5,
    edge_color="#0044cc",
    edge_width=1.5,
    ax=ax2,
)

plt.show()
```
