```python
import numpy as np
import matplotlib.pyplot as plt

def superellipse(a, b, n, num_points=1000):
    """生成超椭圆的参数方程"""
    theta = np.linspace(0, 2 * np.pi, num_points)
    x = a * np.sign(np.cos(theta)) * np.abs(np.cos(theta)) ** (2/n)
    y = b * np.sign(np.sin(theta)) * np.abs(np.sin(theta)) ** (2/n)
    return x, y
def superellipse1(a, b, n, num_points=1000):
    """生成超椭圆的参数方程"""
    theta = np.linspace(0, 2 * np.pi, num_points)
    x = a *np.cos(theta) ** (2/n)
    y = b *np.sin(theta) ** (2/n)
    return x, y

# 参数设置
a, b = 2, 1  # 半轴长度
n_values = [1/12]  # 不同的n值

# 绘制图形
plt.figure(figsize=(10, 6))
for n in n_values:
    x, y = superellipse(a, b, n)
    plt.plot(x, y, label=f'n = {n}')

# 美化图形
plt.title("Superellipses with Different n Values")
plt.xlabel("x")
plt.ylabel("y")
plt.axhline(0, color='gray', linestyle='--', linewidth=0.5)
plt.axvline(0, color='gray', linestyle='--', linewidth=0.5)
plt.grid(True, linestyle=':', alpha=0.7)
plt.legend()
plt.axis('equal')  # 保证比例一致
plt.show()
```

```python
import numpy as np
import plotly.graph_objects as go

def superellipse(a, b, n, num_points=1000):
    """生成超椭圆的参数方程"""
    theta = np.linspace(0, 2 * np.pi, num_points)
    x = a * np.sign(np.cos(theta)) * np.abs(np.cos(theta)) ** (2/n)
    y = b * np.sign(np.sin(theta)) * np.abs(np.sin(theta)) ** (2/n)
    return x, y

# 参数设置
a, b = 2, 1  # 半轴长度
n_values = [1/12]  # 不同的n值

# 创建图形
fig = go.Figure()

# 添加每条超椭圆曲线
for n in n_values:
    x, y = superellipse(a, b, n)
    fig.add_trace(go.Scatter(
        x=x,
        y=y,
        mode='lines',
        name=f'n = {n}',
        line=dict(width=2)
    ))

# 美化图形
fig.update_layout(
    title='Superellipses with Different n Values',
    xaxis_title='x',
    yaxis_title='y',
    xaxis=dict(showgrid=True, gridcolor='lightgray', zeroline=True, zerolinecolor='gray'),
    yaxis=dict(showgrid=True, gridcolor='lightgray', zeroline=True, zerolinecolor='gray'),
    legend_title='n Values',
    hovermode='closest',
    width=800,
    height=600
)

# 显示图形
fig.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from flower import flowers_flower_by_petal_multi
import os

# 创建保存图片的目录
# if not os.path.exists('flower_images'):
#     os.makedirs('flower_images')
count=0
# 生成1到4之间的数列，步长为0.1
rs=np.arange(2.0,2.1,0.1)
ns=np.arange(3,12,1)
ratios=np.arange(0.5,1.8,0.1)
Ms=np.arange(3,5,1)
Ns=np.arange(3,12,1)
# 为每个参数值单独绘制并保存图片
for r in rs:
    for n in ns:
        for ration in ratios:
            for M in Ms:
                for N in Ns:
                    # 创建一个新的图形，设置合适的大小
                    fig, ax = plt.subplots(1, 1, figsize=(8, 8))

                    # 调用绘图函数
                    flowers_flower_by_petal_multi(1, r, n, ration, M, N)

                    # 移除坐标轴、标签、图例等所有额外元素
                    ax.set_xticks([])
                    ax.set_yticks([])
                    ax.set_xlabel("")
                    ax.set_ylabel("")
                    ax.set_title("")
                    ax.axis("off")

                    # 调整布局，确保只保存图形本身
                    plt.tight_layout(pad=0)

                    # 保存图片，使用参数去除所有空白
                    filename = f"../../../images/flower_1_{r:.1f}_{n}_{ration:.1f}_{M}_{N}.png"
                    plt.savefig(
                        filename,
                        bbox_inches="tight",
                        pad_inches=0,
                        dpi=150,  # 适中的DPI保证清晰度但不过大
                        facecolor="white",
                        edgecolor="none",
                    )

                    # 关闭图形以释放内存
                    plt.close(fig)
                    count+=1
                    print(f"已保存: {filename}")

print(f"总共保存了 {count} 张图片到 'flower_images' 目录")
```

```python
import matplotlib.pyplot as plt
import numpy as np
import matplotlib.cm as cm  # 渐变配色库
from cprdspy.CPR_matplotlib.Arcs.arc import *


def rotate_point(point, theta):
    x, y = point
    x_new = x * np.cos(theta) - y * np.sin(theta)
    y_new = x * np.sin(theta) + y * np.cos(theta)
    return (x_new, y_new)


def draw_oval_flower_gradient(
    a=3,
    b=1.5,
    d=0.3,
    n=8,
    alpha=0.8,
    center=(0, 0),
    points=1000,
    center_circle_r=0.2,
    center_circle_color="#ff4d94",
    peak_size=8,
    peak_color="#ffff00",
    peak_edge="#000",
):
    ax = plt.gca()
    ax.set_aspect("equal")

    # 1. 绘制中心圆
    circle = plt.Circle(
        center,
        center_circle_r,
        facecolor=center_circle_color,
        edgecolor="#fff",
        linewidth=0.5,
        alpha=alpha,
        zorder=5,
    )
    ax.add_patch(circle)

    # 2. 生成渐变配色（从浅蓝到深蓝）
    # Blues colormap：0→浅蓝，1→深蓝
    colors = cm.Blues(np.linspace(0.4, 0.9, n))  # 8个花瓣对应8种深浅

    # 3. 循环绘制渐变花瓣+点
    for i in range(n):
        rotation = i * 2 * np.pi / n
        petal_color = colors[i]  # 每个花瓣用不同深浅的蓝色

        # 椭圆花瓣参数计算
        x0 = b / (2 * a) * np.sqrt(4 * a**2 - d**2) / (d / 2)
        beta = np.arctan(x0)
        beta_b1 = np.pi / 2 - beta
        beta_e1 = np.pi / 2 + beta
        beta_b2 = 3 * np.pi / 2 - beta
        beta_e2 = 3 * np.pi / 2 + beta

        # 椭圆弧中心点
        center1 = (center[0], center[1] - d / 2)
        center2 = (center[0], center[1] + d / 2)
        center1_rot = rotate_point(center1, rotation)
        center2_rot = rotate_point(center2, rotation)

        # 绘制渐变椭圆弧（圈）
        oval_arc(
            a,
            b,
            beta_b1,
            beta_e1,
            angle=rotation,
            color=petal_color,
            alpha=alpha,
            center=center1_rot,
            points=points,
            use_degree=False,
        )
        oval_arc(
            a,
            b,
            beta_b2,
            beta_e2,
            angle=rotation,
            color=petal_color,
            alpha=alpha,
            center=center2_rot,
            points=points,
            use_degree=False,
        )

        # 绘制花瓣顶点散点（点）
        petal_peak = (
            center[0] + a * np.cos(rotation),
            center[1] + a * np.sin(rotation),
        )
        ax.scatter(
            petal_peak[0],
            petal_peak[1],
            s=peak_size,
            c=peak_color,
            edgecolor=peak_edge,
            linewidth=0.5,
            alpha=alpha,
            zorder=10,
        )


# 调用渐变版椭圆花
plt.figure(figsize=(8, 8), dpi=120)
plt.axis("off")
plt.gca().set_aspect("equal")
plt.gcf().patch.set_facecolor("#10101a")

draw_oval_flower_gradient(
    a=3,
    b=1.5,
    d=0.3,
    n=8,
    alpha=0.8,
    center_circle_r=0.2,
    center_circle_color="#ff4d94",
    peak_size=8,
    peak_color="#ffff00",
    peak_edge="#000",
)

plt.tight_layout()
plt.show()
```
