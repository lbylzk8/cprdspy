```python
import numpy as np
import matplotlib.pyplot as plt
from flower import flowers_flower_by_petal_multi
import os

# 创建保存图片的目录
# if not os.path.exists('flower_images'):
#     os.makedirs('flower_images')

# 生成1到4之间的数列，步长为0.1
rs=np.arange(1.2,2,0.1)
ns=np.arange(3,12,1)
ratios=np.arange(0.5,1.8,0.1)
Ms=np.arange(3,5,1)
Ns=np.arange(3,12,1)
# 为每个参数值单独绘制并保存图片
for r in rs:
    for n in ns:
        for ration in ratios:
            for M in Ms:
                for N in Ns:
                    # 创建一个新的图形，设置合适的大小
                    fig, ax = plt.subplots(1, 1, figsize=(8, 8))

                    # 调用绘图函数
                    flowers_flower_by_petal_multi(1, r, n, ration, M, N)

                    # 移除坐标轴、标签、图例等所有额外元素
                    ax.set_xticks([])
                    ax.set_yticks([])
                    ax.set_xlabel("")
                    ax.set_ylabel("")
                    ax.set_title("")
                    ax.axis("off")

                    # 调整布局，确保只保存图形本身
                    plt.tight_layout(pad=0)

                    # 保存图片，使用参数去除所有空白
                    filename = f"../../../images/flower_1_{r:.1f}_{n}_{ration:.1f}_{M}_{N}.png"
                    plt.savefig(
                        filename,
                        bbox_inches="tight",
                        pad_inches=0,
                        dpi=150,  # 适中的DPI保证清晰度但不过大
                        facecolor="white",
                        edgecolor="none",
                    )

                    # 关闭图形以释放内存
                    plt.close(fig)

                    print(f"已保存: {filename}")

print(f"总共保存了 {len(sequence)} 张图片到 'flower_images' 目录")
```
