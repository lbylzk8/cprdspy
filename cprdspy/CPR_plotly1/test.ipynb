# CPR_plotly1 模块测试

本 Notebook 用于测试 CPR_plotly1 模块中的各个功能。

```python
import numpy as np
import plotly.graph_objects as go
from cprdspy.CPR_plotly1 import *
```

## 1. 测试 Circles 模块

```python
# 测试 circle 函数
circle_trace = circle_plotly(radius=2, color="#00ff00", center=(0, 0))
fig = go.Figure(data=[circle_trace])
fig.update_layout(
    title="测试 circle_plotly",
    xaxis=dict(scaleanchor="y", scaleratio=1),
    yaxis=dict(scaleanchor="x", scaleratio=1),
    width=500,
    height=500,
)
fig.show()
```

```python
# 测试 concentric_circles 函数
traces = concentric_circles_plotly(center=(0, 0), n=5, radius=1, param=0.3, mode="arithmetic", direction="both")
fig = go.Figure(data=traces)
fig.update_layout(
    title="测试 concentric_circles_plotly（等差）",
    xaxis=dict(scaleanchor="y", scaleratio=1),
    yaxis=dict(scaleanchor="x", scaleratio=1),
    width=500,
    height=500,
)
fig.show()
```

```python
# 测试 ellipse 函数
ellipse_trace = ellipse_plotly(a=3, b=1.5, rotation=30, color="#ff00ff")
fig = go.Figure(data=[ellipse_trace])
fig.update_layout(
    title="测试 ellipse_plotly（旋转 30 度）",
    xaxis=dict(scaleanchor="y", scaleratio=1),
    yaxis=dict(scaleanchor="x", scaleratio=1),
    width=500,
    height=500,
)
fig.show()
```

## 2. 测试 Arcs 模块

```python
# 测试 arc 函数
arc_trace = arc_plotly(r=2, angle1=0, angle2=90, rotation=0, color="#00ffff")
fig = go.Figure(data=[arc_trace])
fig.update_layout(
    title="测试 arc_plotly（0-90 度圆弧）",
    xaxis=dict(scaleanchor="y", scaleratio=1),
    yaxis=dict(scaleanchor="x", scaleratio=1),
    width=500,
    height=500,
)
fig.show()
```

```python
# 测试 arc_point 函数
arc_trace = arc_point_plotly(center=(0, 0), point1=(2, 0), point2=(0, 2), color="#ffff00")
fig = go.Figure(data=[arc_trace])
fig.update_layout(
    title="测试 arc_point_plotly（通过两点画弧）",
    xaxis=dict(scaleanchor="y", scaleratio=1),
    yaxis=dict(scaleanchor="x", scaleratio=1),
    width=500,
    height=500,
)
fig.show()
```

```python
# 测试 oval_arc 函数
oval_arc_trace = oval_arc_plotly(a=3, b=1.5, angle1=0, angle2=180, angle=30, color="#ff9900")
fig = go.Figure(data=[oval_arc_trace])
fig.update_layout(
    title="测试 oval_arc_plotly（椭圆弧）",
    xaxis=dict(scaleanchor="y", scaleratio=1),
    yaxis=dict(scaleanchor="x", scaleratio=1),
    width=500,
    height=500,
)
fig.show()
```

## 3. 测试 Flowers 模块

```python
# 测试 flower 函数（单层花）
traces = flower_plotly(R=2, r=2, n=4, N=8, color="#ff66cc")
fig = go.Figure(data=traces)
fig.update_layout(
    title="测试 flower_plotly（单层花，8 瓣）",
    xaxis=dict(scaleanchor="y", scaleratio=1),
    yaxis=dict(scaleanchor="x", scaleratio=1),
    width=500,
    height=500,
)
fig.show()
```

## 6. 综合测试 - 组合图案

```python
# 组合多种元素
all_traces = []

# 添加同心圆
all_traces.extend(concentric_circles_plotly(center=(0, 0), n=3, radius=0.5, param=0.5, color="#336699"))

# 添加花朵
all_traces.extend(flower_plotly(R=1.5, r=1.5, n=4, N=12, color="#ff6699"))

# 添加螺旋
all_traces.extend(n_spiral_rotate_plotly(n=4, cyc=1, color="#99ff99", alpha=0))

fig = go.Figure(data=all_traces)
fig.update_layout(
    title="综合测试 - 组合图案",
    xaxis=dict(scaleanchor="y", scaleratio=1, showgrid=False, zeroline=False),
    yaxis=dict(scaleanchor="x", scaleratio=1, showgrid=False, zeroline=False),
    plot_bgcolor="#10101a",
    paper_bgcolor="#10101a",
    width=700,
    height=700,
)
fig.show()
```
