```python
import numpy as np
from cprdspy.CPR_matplotlib import flower_mpl
import matplotlib.pyplot as plt

# 1. 初始化画布（核心自定义项）
plt.figure(
    figsize=(10, 10),  # 画布大小（宽×高，单位：英寸），越大越清晰
    dpi=150,  # 分辨率（ dots per inch），300=高清，120=普通
    facecolor="#0a0a12",  # 画布背景色（和下面的patch颜色保持一致）
)

# 2. 自定义坐标轴
ax = plt.gca()  # 获取当前坐标轴
ax.set_aspect("equal")  # 强制等比例，圆/椭圆不拉伸（CPR艺术必开）
ax.axis("off")  # 隐藏坐标轴（推荐，让图案更纯净）
# 若想显示坐标轴（调试用）：ax.axis("on")，可进一步自定义：
# ax.set_xlim(-5, 5)  # x轴范围
# ax.set_ylim(-5, 5)  # y轴范围
# ax.set_xticks([])  # 隐藏x轴刻度
# ax.set_yticks([])  # 隐藏y轴刻度

# 3. 自定义画板背景（和figure的facecolor呼应）
plt.gcf().patch.set_facecolor("#0a0a12")  # gcf()=获取当前画布，patch=背景层
ax.set_facecolor("#0a0a12")  # 坐标轴背景色（避免白边）

# 4. 绘制花型（自定义花的参数）
# flower_mpl()

# 5. 保存图片（可选，自定义保存参数）
# plt.savefig(
#     "我的CPR自定义图案.png",  # 保存路径+文件名
#     dpi=300,  # 保存分辨率（建议300，适合做壁纸/海报）
#     bbox_inches="tight",  # 去除图片白边（必加！）
#     facecolor="#0a0a12",  # 保存时保留背景色
#     edgecolor="none",  # 去除图片边框
# )

# 6. 显示图片
# plt.show()
```
